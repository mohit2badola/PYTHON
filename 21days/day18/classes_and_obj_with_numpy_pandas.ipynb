1. NumPy inside Class (Array Operations)

```python
import numpy as np
```

```python
class Arrayanalyser:
    def __init__(self,data):
        self.arr=np.array(data)

    def show(self):
        print("Array : ",self.arr)

    def stats(self):
        print("Sum : ", np.sum(self.arr))
        print("Mean : ", np.mean(self.arr))
        print("Std : ", np.std(self.arr))
```

```python
a=Arrayanalyser([1,2,3,4,5,6])
```

```python
a.show()
a.stats()
# Arrayanalyser([1,2,3,4,5,6]).stats  # need magic function for this 
```

2. Using __str__ with NumPy

```python
class arraydisplay:
    def __init__(self,data):
        self.arr=np.array(data)

    def __str__(self):
        return f"My array is : {self.arr}"

```

```python
a=arraydisplay([10,20,30,40])
print(a)

```

3. Overloading + for Arrays

```python
class arrayadd:
    def __init__(self,data):
        self.arr=np.array(data)

    def __add__(self,other):
        return self.arr+ other.arr

```

```python
a1=arrayadd([1,2,3])
a2=arrayadd([1,2,3])

print(a1+a2)
```

  Q4. Student Data Manager

```python
import pandas as pd
```

```python
class studentdata:
    def __init__(self):
        self.df=pd.DataFrame({
            "Name" : ['A','B','C'],
            "Marks" : [90,75,60]

        })

    def show(self):
        print(self.df)

    def average(self):
        print(f"Average : ",self.df["Marks"].mean())

    def topper(self):
        print(self.df[self.df["Marks"]==self.df["Marks"].max()])
```

```python
s=studentdata()
s.show()
s.average()
s.topper()
```
