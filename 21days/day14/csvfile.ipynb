## BASICS OF CSV FILE 

Reading into csv file

```python
import csv
with open ("Data.txt", "r") as f:
    reader=csv.reader(f)
    for row in reader:
        print(row)


```

writing in csv file

```python
import csv
with open ("Data.txt", "w") as f:
    writer=csv.writer(f)
    writer.writerow(["Welccome","everyone"])
    writer.writerow([["Welccome","everyone"],[25,30]])



```
