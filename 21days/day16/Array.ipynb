## Numpy /Pandas

Reshape an array

```python
import numpy as np
```

```python
arr=np.array([1,2,3,4,5,6])

print(arr.reshape(2,3))
```

```python
arr=np.array([1,2,3,4,5,6])

print(arr.reshape(3,2))
```

```python
arr=np.array([1,2,3,4,5,6])

print(arr.reshape(6,1))
```

```python
arr=np.array([1,2,3,4,5,6])

print(arr.reshape(1,6))
```

```python
arr=np.array([1,2,3,4,5,6])

print(arr.reshape(4,1))
```

Important Functions 

```python
arr=np.array([1,2,3,4,5,6])
```

```python
print(np.sum(arr))
print(np.mean(arr))
print(np.std(arr))
print(np.amax(arr))
```

```python
print(np.random.rand(3))
print(np.random.randint(1,10,5))
```

Attributes

```python
arr=np.array([[1,2,3],[4,5,6]])

print(arr.shape)
print(arr.ndim)
print(arr.size)
```

Data Frame Creation

```python
df=pd.DataFrame({
    "Name": ["A","B"],
    "Marks" : [90,80]
})

print(df)
```

Column Operations

```python
print(df["Marks"])
```

```python
df["Age"]=[20,22]
df
```

```python
df.drop("Age",axis=1, inplace=True)
df
```

Filtering

```python
print(df[df["Marks"]>85])
```

Missing Data

```python
df["Marks"][0]=None
print(df.isnull())
print(df.fillna(0))
df
```

Aggregation

```python
df.to_csv("data.csv")
new_df=pd.read_csv("data.csv")
print(new_df)
```
