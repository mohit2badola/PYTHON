## Function Overloading 

Same function name, different behavior depending on arguments

```python
class Mathsops:
    def add(self,a,b=5,c=0):
        return a+b+c
a=Mathsops()
print(a.add(3,7))
print(a.add(3,6,3))
print(a.add(3))

```

Using *args Any number of arguments

```python
class Mathsops:
    def add(self,*args):
        return sum(args)
a=Mathsops()
print(a.add(3,7))
print(a.add(3,6,3))
print(a.add(3,3,3234,6,4,3,3,4,5,6))

```

## Operator Overloading 

Giving new meaning to operators like +, -, *

```python
class Number:
    def __init__(self,value):
        self.value=value

    def __add__(self,other):
        return self.value+other.value


n1=Number(10)
n2=Number(56)
print(type(n1))
print(n1+n2)
```

overloading with multiple objects

```python
class Number:
    def __init__(self,value):
        self.value=value

    def __add__(self,other):
        return Number(self.value+other.value)

    def __str__(self):
        return str(self.value)


n1=Number(10)
n2=Number(56)
n3=Number(562)
n4=Number(562)

print(n1+n2+n3+n4)
```

Operator Overloading for Subraction

```python
class Number:
    def __init__(self,value):
        self.value=value

    def __sub__(self,other):
        return Number(self.value-other.value)

    def __str__(self):
        return str(self.value)


n1=Number(10)
n2=Number(56)
n3=Number(52)
n4=Number(2)

print(n1-n2-n3-n4)
```

Operator Overloading for Multiplication

```python
class Number:
    def __init__(self,value):
        self.value=value

    def __mul__(self,other):
        return Number(self.value*other.value)

    def __str__(self):
        return str(self.value)


n1=Number(10)
n2=Number(56)
n3=Number(562)
n4=Number(562)

print(n1*n2*n3*n4)
```
