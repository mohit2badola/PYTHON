```python
import json

class ConvertToJSON:
    
    def __init__(self, data):
        self.data = data

    def convert(self):
        new_data = {}
        
        for key, value in self.data.items():
            new_data[str(key)] = value
        
        json_data = json.dumps(new_data, indent=4)
        return json_data


records = {
    101: {"name": "Aman", "marks": 78},
    102: {"name": "Neha", "marks": 91}
}

obj = ConvertToJSON(records)

result = obj.convert()

print(result)
```

```python
class EncryptSAP:
    def __init__(self, sapid):
        self.sapid = sapid

    # 🔹 Encrypt function
    def encrypt(self):
        encrypted = ""
        for ch in self.sapid:
            encrypted += chr((ord(ch) + 5) % 126)
        return encrypted

    # 🔹 Decrypt function (reverse of encrypt)
    def decrypt(self, encrypted_text):
        decrypted = ""
        for ch in encrypted_text:
            decrypted += chr((ord(ch) - 5) % 126)
        return decrypted

    # 🔹 Add function (+1 shift)
    def __add__(self, other):
        new_sapid = ""
        for ch in self.sapid:
            new_sapid += chr((ord(ch) + 1) % 126)
        return EncryptSAP(new_sapid)

    # 🔹 String output
    def __str__(self):
        return f"Encrypted SAPID: {self.encrypt()}"


# 🔹 Main Program
sap = input("Enter SAP ID: ")

obj = EncryptSAP(sap)

# Encrypt
encrypted_text = obj.encrypt()
print("Encrypted:", encrypted_text)

# Decrypt
decrypted_text = obj.decrypt(encrypted_text)
print("Decrypted:", decrypted_text)

# +1 operation
new_obj = obj + 1
print("After +1 operation:", new_obj)
```
