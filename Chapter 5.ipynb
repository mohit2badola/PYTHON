Make a dictionary to dictionary to jason to jason IMPORTANT QUESTION

```python
import json

class ConvertToJSON:
    
    def __init__(self, data):
        self.data = data

    def convert(self):
        new_data = {}
        
        for key, value in self.data.items():
            new_data[str(key)] = value
        
        json_data = json.dumps(new_data, indent=4)
        return json_data


records = {
    101: {"name": "Aman", "marks": 78},
    102: {"name": "Neha", "marks": 91}
}

obj = ConvertToJSON(records)

result = obj.convert()

print(result)
```

```python
class Student:
    def set_details(self, name, roll):
        self.name = name
        self.roll = roll

    def display(self):
        print(self.name, self.roll)

s1 = Student()
s1.set_details("Aman", 101)
s1.display()
```
