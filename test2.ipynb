```python
class product:
    def __init__(self, name,price):
        self.name=name
        self.price=price

    def __str__(self):
        return f"{self.name} , {self.price}"
    def __add__(self,other):
        return f"{self.price+other.price}"

    def __eq__(self,other):
         return f"{self.price==other.price}"

p1 = product("Laptop", 50000)
p2 = product("Phone", 30000)
p3 = product("Tablet", 50000)


print(p1)
print(p1 + p2)
print(p1 == p3)
print(p1 == p2)
```
